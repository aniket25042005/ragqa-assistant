```python
import logging
logging.getLogger("pypdf").setLevel(logging.ERROR)

from pypdf import PdfReader

reader = PdfReader("../media/documents/Environmental_Pollution.pdf") 
text = ""
for page in reader.pages:
    text += page.extract_text() or ""

print(len(text))
print(text[:500])
```

# Chunking 

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=700,
    chunk_overlap=100
)
chunks = splitter.split_text(text)

print(f"Number of chunks: {len(chunks)}")
print(chunks[0])
print("---")
print(chunks[1])
```

# Embeddings

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode(chunks)

print(embeddings.shape) 
print(embeddings[0][:10]) 
```

# Similarity Search

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

query = "What are the effects of air pollution?"  # adjust to something relevant to your PDF
query_embedding = model.encode([query])

similarities = cosine_similarity(query_embedding, embeddings)[0]

top_idx = np.argsort(similarities)[::-1][:3]  # top 3 matches
for idx in top_idx:
    print(f"Score: {similarities[idx]:.3f}")
    print(chunks[idx])
    print("---")
```
