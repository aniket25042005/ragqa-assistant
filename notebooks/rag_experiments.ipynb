```python
import logging
logging.getLogger("pypdf").setLevel(logging.ERROR)

from pypdf import PdfReader

reader = PdfReader("../media/documents/Environmental_Pollution.pdf") 
text = ""
for page in reader.pages:
    text += page.extract_text() or ""

print(len(text))
print(text[:500])
```

# Chunking 

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=700,
    chunk_overlap=100
)
chunks = splitter.split_text(text)

print(f"Number of chunks: {len(chunks)}")
print(chunks[0])
print("---")
print(chunks[1])
```

# Embeddings

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode(chunks)

print(embeddings.shape) 
print(embeddings[0][:10]) 
```

# Similarity Search

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

query = "What are the effects of air pollution?"  
query_embedding = model.encode([query])

similarities = cosine_similarity(query_embedding, embeddings)[0]

top_idx = np.argsort(similarities)[::-1][:3]  
for idx in top_idx:
    print(f"Score: {similarities[idx]:.3f}")
    print(chunks[idx])
    print("---")
```

# Storing in FAISS

```python
import faiss
import numpy as np

embeddings_np = np.array(embeddings).astype('float32')

dimension = embeddings_np.shape[1]  
index = faiss.IndexFlatIP(dimension) 

faiss.normalize_L2(embeddings_np)
index.add(embeddings_np)

print(f"Number of vectors in index: {index.ntotal}")
```

## Testing retrieval through FAISS instead of cosine similarity

```python
query = "What are the effects of air pollution?"
query_embedding = model.encode([query]).astype('float32')
faiss.normalize_L2(query_embedding)

k = 3 
distances, indices = index.search(query_embedding, k)

for i, idx in enumerate(indices[0]):
    print(f"Score: {distances[0][i]:.3f}")
    print(chunks[idx])
    print("---")
```

```python
faiss.write_index(index, "../qa/document_index.faiss") #saving index to disk
```
